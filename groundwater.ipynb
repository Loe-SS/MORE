# Groundwater

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df = pd.read_csv('data/groundwater.csv', delimiter=';')
df['meetwaarde (m NAP)'] = df['meetwaarde (m NAP)'] / 100
df['date'] = pd.to_datetime(df['datum'] + ' ' + df['tijd'], dayfirst=True)
df = df.set_index('date')
df = df.drop(columns=['datum', 'tijd'])
df = df.resample('D').mean()
```

```python
display(df)
```

```python
plt.figure(figsize=(15,5))

plt.plot(df.index, df['meetwaarde (m NAP)'], label='Groundwater Level (m NAP)', color='blue')
plt.xlabel('Date')
plt.ylabel('Groundwater Level (m NAP)')
plt.axhline(8.33, color = 'r', linestyle = '--', label='Ground level')
plt.legend()
plt.tight_layout()
plt.grid()
```
